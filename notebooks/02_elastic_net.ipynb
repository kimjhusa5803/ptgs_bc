# 02 — Elastic-net baseline (tuning)

The reference method: build the PTGS with elastic net, tuning **alpha** (penalty) and
**l1_ratio** (L1↔L2 mix) in the inner CV. `l1_ratios` is widened toward ridge because
correlated TWAS genes benefit from elastic net's grouping (pure lasso arbitrarily drops one of
a correlated set). The plot shows how the inner-CV error varies across the tuned parameters.

## Setup + data

```python
import ptgs_bc as ptgs
from ptgs_bc import simulate_dataset, ElasticNetBuilder, nested_cv, enet_cv_path, plot_enet_path
%matplotlib inline
ds, _ = simulate_dataset(n_samples=400, n_genes=100, n_causal=15, family="gaussian", seed=0)
```

## Fit + selected hyperparameters

```python
b = ElasticNetBuilder()   # l1_ratios widened to (0.05 .. 1.0)
bundle = b.fit(ds, seed=0)
print("chosen:", bundle.extras["chosen"], "| non-zero genes:", bundle.extras["n_nonzero"])
print("nested-CV:", nested_cv(b, ds, outer_k=5, seed=0))
```

## Tuning surface — inner-CV error vs the penalty

One line per `l1_ratio`; the x-axis is log10(alpha). The minimum of the lowest curve is the
selected (alpha, l1_ratio). This is the elastic-net "optimization variation across parameters".

```python
path = enet_cv_path(ds)
ax = plot_enet_path(path)
ax.figure
```
